## Exercise 6: Rectangle Calculator

Write a program that asks the user for the length and width of a rectangle.

Calculate and display:

- The area
- The perimeter

### Formulas

$Area = length \times width$

$Perimeter = 2(length + width)$


### Example Run

```text
Enter the length: 10
Enter the width: 5

Area: 50
Perimeter: 30
```

````python
length = int(input("Enter the length: "))
width = int(input("Enter the width: "))
area = length * width
perimeter = 2 * (length + width)

print(f"Area: {area}\nPerimeter: {perimeter}")
````

## Exercise 9: Restaurant Bill

Write a program that asks the user for the cost of a meal.

Use these named constants:

```python
TAX_RATE = 0.0825
TIP_RATE = 0.20
```

Calculate:

- The meal cost
- The tax
- The tip
- The total cost

Display all four values.

````python
TAX_RATE = 0.0825
TIP_RATE = 0.20

meal_cost = float(input("Enter the cost of the meal: "))
tax = meal_cost * TAX_RATE
tip = meal_cost * TIP_RATE
total_cost = meal_cost + tax + tip
print(f"Meal Cost: ${meal_cost}\nTax: ${tax:.2f}\nTip: ${tip:.2f}\nTotal Cost: ${total_cost:.2f}")
````

## Exercise 11: Temperature Converter

Write a program that asks the user for a temperature in Fahrenheit.

Convert the temperature to Celsius using:

\[
C = \frac{5}{9}(F - 32)
\]

Display the Celsius temperature rounded to **one decimal place**.

### Example Run

```text
Enter temperature in Fahrenheit: 80

80.0°F is 26.7°C.
```

````python
f_temp = float(input("Enter temperature in Fahrenheit: "))
c_temp = (f_temp - 32) * 5 / 9
print(f"{f_temp}°F is {c_temp:.1f}°C.")
````
