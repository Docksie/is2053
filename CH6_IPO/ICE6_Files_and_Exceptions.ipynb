## Exercise 1: Write to a File

Write a program that creates a file named `message.txt`.

## Exercise 3: Write and Read User Data

Ask the user to enter:

- Their name
- Their favorite food
- Their favorite number

Write the three values to a file named `student_info.txt`, with each value on its own line.

Then open the file again and display its contents.

Use a `with` statement for **both** writing and reading.

````python
with open("student_info.txt", "w") as f:
    f.writelines(input("Enter your name: ") + '\n')
    f.writelines(input("Enter your favorite food: ") + '\n')

    f.writelines(input("Enter your favorite number: ") + '\n')

with open ("student_info.txt", "r") as f:
    print(f.read())
````

## Exercise 4: Reading a File with a Loop

Create a file named `scores.txt` (you can do this manually).

`scores.txt` should contain these scores:

```text
88
92
75
100
84
```

Then write a program that reads the file **one line at a time using a `for` loop**.

Display each score.

### Hint

Each line read from the file is a string and contains a newline character.

You can remove the newline using:

```python
line.strip()
```

````python
with open("scores.txt", 'w') as f:
    f.write("88\n92\n75\n100\n84")
````

## Exercise 5: Calculate an Average from a File

Using the same `scores.txt` file, write a program that:

1. Reads each score using a `for` loop.
2. Converts each score to an integer.
3. Keeps a running total.
4. Counts how many scores were read.
5. Calculates and displays the average.

### Example Output

```text
Average score: 87.8
```

````python
total = 0
count = 0
with open('scores.txt', 'r') as f:
    for line in f:
        current = int(line.strip())
        total += current
        count += 1
print(f'Average score: {total/count}')
````

## Exercise 6: Appending to a File

Ask the user to enter a new score.

Append the new score to `scores.txt` without deleting the scores that are already there.

Then read and display the entire file to verify that the new score was added.

Use append mode (`"a"`).

````python
with open("scores.txt", 'a') as f:
    f.write('\n' + input("Enter a new score: "))

with open("scores.txt", 'r') as f:
    print(f.read())
````

## Exercise 7: Simple Records

Create a file named `students.txt`.

Write the following student records to the file:

```text
Alice,85
Bob,92
Charlie,78
```

Then read the file one line at a time.

For each line:

1. Remove the newline using `.strip()`.
2. Split the line at the comma using `.split(",")`.
3. Display the student's name and score.

### Example Output

```text
Alice scored 85
Bob scored 92
Charlie scored 78
```

````python
with open("students.txt", 'w') as f:
    f.write("Alice,85\nBob,92\nCharlie,78\n")

with open('students.txt', 'r') as f:
    for line in f:
        line = line.strip().split(',')
        print(f'{line[0]} scored {line[1]}')
````

## Exercise 8: Handle a Missing File

Write a program that attempts to open and read a file named:

```text
grades.txt
```

Use `try` and `except` so that the program does not crash if the file does not exist.

If the file is missing, display:

```text
Error: grades.txt was not found.
```

````python
try:
    with open("grades.txt", 'r') as f:
        print(f.read())
except FileNotFoundError:
    print("grades.txt was not found.")
````

## Exercise 9: Handle Invalid Numeric Data

Use the following starter code.

The program asks the user for a number and attempts to convert it to an integer.

Add exception handling so that:

- If the user enters a valid integer, display the number.
- If the user enters something that cannot be converted to an integer, display:

```text
Please enter a valid whole number.
```

````python
user_input = input("Enter a whole number: ")

# Add your try/except code below.
try:
    user_input = int(user_input)
    print(user_input)
except ValueError:
    print("Please enter a valid whole number.")
````

## Exercise 10: Multiple Exceptions

Write a program that:

1. Asks the user for the name of a file.
2. Opens the file using a `with` statement.
3. Reads the first line from the file.
4. Converts that line to an integer.

Handle these two possible errors separately:

- `FileNotFoundError`
- `ValueError`

Display an appropriate message for each error.

If no error occurs, display the number that was read from the file.

````python
file_name = input("Enter the file name (include the file extension. i.e. '.txt'): ")
try:
    with open(file_name, 'r') as f:
        try:
            line = f.readline().strip()
            print(int(line))
        except ValueError:
            print("This line doesn't consist of integers.")
except FileNotFoundError:
    print("This file doesn't exist.")

````
